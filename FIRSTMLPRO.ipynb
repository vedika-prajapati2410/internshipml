```python
import tensorflow as tf
import numpy as np
from tensorflow import keras
model = tf.keras.Sequential([keras.layers.Dense(units=1, input_shape=[1])])
#units=1 tells model that output should be one or any no. depending on what you put
#input_shape=[1] tells how many inputs to expect to the model
model.compile(optimizer='sgd', loss='mean_squared_error')
#stochastic gradient descent (sgd)
xs = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-2.0, 1.0, 4.0, 7.0, 10.0, 13.0], dtype=float)
model.fit(xs, ys, epochs=500)
# model is like your puppy. It's the thing you're trying to train..fit() is like the actual training sessions you have with your puppy. This is when the learning happens.# xs is like the balls you throw for the puppy. These are the things you want the puppy to learn to fetch.ys is like you saying "Fetch!"
#epochs= how many times it will run
print(model.predict([10.0]))
```
